### 💬 Mostrar números de forma legible

Un resultado como `1234567.891` no se le muestra a nadie así. En un informe, en una reunión o en un
tablero, **el número se presenta**. Dentro de un f-string, después de los dos puntos `:` se indica
**cómo** formatear:

| Formato | Qué hace | Ejemplo |
|---|---|---|
| `:.2f` | 2 decimales | `1234567.89` |
| `:,.0f` | Separador de miles, sin decimales | `1,234,568` |
| `:,.2f` | Miles **y** 2 decimales | `1,234,567.89` |
| `:.1%` | Como porcentaje | `12.3%` |
| `:+,.2f` | Fuerza el signo (útil para variaciones) | `+1,234,567.89` |
| `:>10` | Alineado a la derecha en 10 espacios | `      hola` |

Lo vas a usar en **toda** la materia para que las salidas se lean.

````python
facturacion = 1234567.891
margen = 0.1234
variacion = -0.0876

print("Sin formato :", facturacion)
print(f"2 decimales : {facturacion:.2f}")
print(f"Con miles   : {facturacion:,.0f}")
print(f"Completo    : $ {facturacion:,.2f}")
print(f"Porcentaje  : {margen:.1%}")
print(f"Variación   : {variacion:+.1%}")   # el + siempre muestra el signo

# Alinear en columnas: muy útil para que una tabla impresa se lea
print()
productos = ["Notebook", "Mouse", "Monitor"]
precios = [890000.5, 12500.75, 340000]
for nombre, precio in zip(productos, precios):
    print(f"{nombre:<12} $ {precio:>12,.2f}")
````

#### 🎭 Formatear **no** es redondear

Esta es la distinción que más problemas trae cuando el número después se usa para calcular:

| | Qué hace | El valor guardado |
|---|---|---|
| `f"{x:.2f}"` | Cambia **cómo se ve** | Queda **intacto** |
| `round(x, 2)` | Cambia **el valor** | Queda **redondeado** |

Regla práctica: **calculá con todos los decimales y redondeá solo al final, para mostrar.**
Si redondeás en cada paso intermedio, los errores se van sumando y el total no cierra.

````python
precio_unitario = 1234.5678

# Formatear: solo cambia la presentación
print(f"Formateado : {precio_unitario:.2f}   ← se ve con 2 decimales")
print("El valor sigue siendo:", precio_unitario)

# Redondear: cambia el número de verdad
precio_redondeado = round(precio_unitario, 2)
print("\nRedondeado :", precio_redondeado, "  ← el valor cambió")

# Por qué importa: 1000 unidades facturadas
print("\nCon el valor completo  : $", f"{precio_unitario * 1000:,.2f}")
print("Con el valor redondeado: $", f"{precio_redondeado * 1000:,.2f}")
print("Diferencia             : $", f"{(precio_unitario - precio_redondeado) * 1000:,.2f}")
````

### ✅ Booleans

**Los booleans** toman dos valores posibles (prestá atención al uso de mayúsculas):
* `True`
* `False`

Estos se pueden usar para evaluar condiciones, por ejemplo, al comparar el valor de diferentes variables. Específicamente, podés crear condiciones usando uno de los siguientes operadores de comparación:
* `==` es igual (recuerde que el operador `=` representa una asignación en Python)
* `!=` no es igual a
* `>=` mayor o igual a
* `<=` menor o igual a
* `is` y `is not` también producen valores booleanos; sin embargo, requieren coincidencias de identidad exactas, a diferencia de `==` y `!=`.

Tené en cuenta que muchas otras funciones y sintaxis devuelven valores booleanos. Por ejemplo, existe la función ``` isinstance(x, var_type) ``` que devuelve el valor booleano correspondiente si la variable x es equivalente al tipo especificado. Verificá siempre las salidas de las funciones para gestionar sus tipos de datos eficazmente.

"True" también corresponde al valor "1", mientras que "False" corresponde a "0".


````python
e = 5

f = 5.00

print(e>f, e==f,  e>=f, e!=f)
````

````python
print(e is f) # Tené en cuenta que, dado que e no es exactamente lo mismo que f, "es" devuelve False.
````

````python
c = "cualquier texto entre comillas simples o dobles es string"
d = 'cualquier texto entre comillas simples o dobles es string'

print(c == d) # Como se mencionó, aunque el segundo usa comillas simples, es equivalente en Python.
````

````python
print(isinstance(c, str)) # Muchas funciones pueden devolver valores booleanos como esta, que verifica si la variable c es string (usá la forma corta str)
````

````python
print(True == 1, False == 0) # El valor booleano de Verdadero es igual a 1; el valor booleano de Falso es igual a 0.
# Esto será útil para variables categóricas.
````

índice más `:`, que se lee como _"hasta, pero sin incluir"_.

Por ejemplo, `2:10` significa _desde el elemento con índice 2 hasta el 10, sin incluir el 10_.
Si no ponés inicio, Python arranca del principio; si no ponés final, va hasta el último.

````python
stock_deposito[0:2]   # del primer elemento (índice 0) hasta el tercero, sin incluirlo
````

````python
stock_deposito[2:]    # del tercer elemento hasta el final. Si fuera [:2], iría del principio al índice 2 sin incluirlo
````

#### 📊 Resumir una lista sin recorrerla

Python trae funciones que responden preguntas sobre una lista entera de una sola vez.
**Son la versión mínima de lo que después haremos con pandas.**

| Función | Qué devuelve |
|---|---|
| `len(lista)` | Cuántos elementos tiene |
| `sum(lista)` | La suma de todos |
| `max(lista)` / `min(lista)` | El mayor / el menor |
| `sorted(lista)` | Una **copia** ordenada |
| `lista.sort()` | Ordena **la lista original** y devuelve `None` |

````python
ventas = [420, 145, 610, 480, 190, 130]

print("Cantidad de productos:", len(ventas))
print("Total vendido        :", sum(ventas))
print("El que más vendió    :", max(ventas))
print("El que menos vendió  :", min(ventas))
print("Promedio             :", sum(ventas) / len(ventas))
````

````python
# sorted() vs .sort() — una diferencia que confunde mucho
original = [420, 145, 610]

copia_ordenada = sorted(original)          # devuelve una lista NUEVA
print("sorted() devuelve:", copia_ordenada)
print("y el original queda intacto:", original)

resultado = original.sort()                 # ordena EN EL LUGAR
print("\n.sort() devuelve:", resultado, "  ← ¡None! no devuelve la lista")
print("pero el original cambió:", original)
````

> ⚠️ **El error clásico:** escribir `lista = lista.sort()`. Como `.sort()` devuelve `None`,
> tu variable queda valiendo `None` y todo lo que sigue falla. Si querés una copia ordenada, usá
> `sorted()`.

### 🔒 Tuplas

Una **tupla** es un contenedor ordenado, igual que una lista, con **una diferencia fundamental:
no se puede modificar**. Una vez creada, no le podés cambiar, agregar ni borrar elementos.

En lugar de corchetes, se escriben entre **paréntesis**. Si tenés un solo elemento, alcanza con
agregarle una coma para que sea una tupla.

Esa rigidez no es un defecto: es **una garantía**. Cuando algo no debería cambiar nunca —el
encabezado de un reporte, las coordenadas de una sucursal, los datos de una operación ya cerrada—
guardarlo en una tupla evita que se modifique por accidente tres pantallas más abajo.

````python
encabezado_reporte = ('Sucursal', 'Producto', 'Unidades')

type(encabezado_reporte)
````

````python
encabezado_reporte[2]   # la indexación funciona igual que en las listas
````

````python
# encabezado_reporte[2] = 'Importe'  # ← si lo descomentás, falla: las tuplas no se pueden modificar
````

````python
# Se puede "desarmar" una tupla asignando cada elemento a una variable distinta
columna_1, columna_2, columna_3 = encabezado_reporte
print('El reporte tiene las columnas', columna_1, ',', columna_2, 'y', columna_3)
````

````python
# del encabezado_reporte[2]   # ← tampoco se pueden borrar elementos
````

````python
encabezado_lista = list(encabezado_reporte)  # una tupla NO se puede cambiar, pero sí convertir en lista

encabezado_lista   # fijate los corchetes: ahora es una lista y se puede modificar
````

````python
columna_extra = 'Importe',   # con la coma alcanza para que sea una tupla de un solo elemento
encabezado_reporte + columna_extra   # sumar tuplas las combina en una nueva
````

## 🔀 Control de flujo

Para manejar el **control de flujo** de un programa necesitamos dos estructuras básicas:
**condiciones** y **ciclos**.

- Las **condiciones** evalúan si algo se cumple y, según la respuesta, siguen por un camino o por otro.
- Los **ciclos** repiten una acción sobre cada elemento de un conjunto, o hasta que se cumpla una condición.

Con estas dos cosas ya se puede escribir prácticamente cualquier programa.

### 🔀 Condicionales con `if`

Un condicional tiene esta forma:

```
if condición == True:
    acción
else:
    otra acción
```

Si hay varias condiciones, se amplía con `elif` (las veces que haga falta):

```
if condición_1 == True:
    acción
elif condición_2 == True:
    acción
elif condición_3 == True:
    acción
...
else:
    acción
```

> 💡 Ojo con los **dos puntos** al final de cada línea y con la **sangría** (los espacios de la línea
> de abajo). En Python la sangría no es decoración: es lo que le dice qué acción pertenece a qué
> condición.

````python
if 1 > 2:
    print('¡lógico!')
else:
    print('no es lógico :(')
````

````python
if 2 > 1:
    print('¡lógico!')
else:
    print('no es lógico :(')
````

### 🔁 Ciclos (`for`)

Los **ciclos** recorren una secuencia de elementos (por ejemplo, una lista) y hacen algo con cada uno.

El ciclo `for` es el más común. Tiene esta forma:

```
for nombre_temporal in secuencia:
    ... nombre_temporal ...
```

Lo que hace Python es:

1. Toma el **primer** elemento de la secuencia y le pone temporalmente la etiqueta `nombre_temporal`.
2. Ejecuta la acción de abajo usando esa etiqueta.
3. Pasa al **siguiente** elemento y repite, hasta terminar la secuencia.

> 💡 El nombre `nombre_temporal` lo elegís vos, y conviene que diga algo: si recorrés una lista de
> sucursales, llamalo `sucursal`. El código se lee solo.

Eso hace que buscar sea directo: en vez de acordarte en qué posición estaba el dato, lo pedís

Si pensás las claves como **nombres de columna** y los valores como el contenido de esa columna,
ya estás viendo por qué los diccionarios se convierten tan fácil en las tablas que vamos a usar

Se escriben así:

mi_diccionario = {'clave_1': valor_1, 'clave_2': valor_2}

### ⚖️ Tupla o diccionario: ¿cuál uso?

Ya viste los dos y es **el momento exacto** en que se mezclan. Los dos guardan varias cosas juntas,
pero responden preguntas distintas:

| | 🔒 Tupla | 📖 Diccionario |
|---|---|---|
| **Cómo accedés** | Por **posición**: `t[0]` | Por **clave**: `d['sucursal']` |
| **Se puede modificar** | ❌ Nunca | ✅ Sí |
| **Tiene orden** | ✅ Sí, es lo único que la ordena | Conserva el de inserción, pero no se usa para buscar |
| **Cuándo se usa** | Un dato fijo, con posiciones que ya sabés qué significan | Un dato con **nombre**, que puede cambiar o crecer |
| **Se parece a** | Una **fila** de una tabla | Un **registro** con sus etiquetas |

La pregunta que lo resuelve: **¿te acordás de memoria qué hay en la posición 2?**
Si la respuesta es no, usá un diccionario.

````python
# La MISMA información guardada de las dos formas

# Como tupla: compacta, pero hay que recordar qué significa cada posición
venta_tupla = ('Sucursal Centro', 'monitores', 12, 340000)
print("Unidades vendidas:", venta_tupla[2], "  ← ¿estabas seguro de que era la posición 2?")

# Como diccionario: cada dato dice qué es
venta_dicc = {
    'sucursal': 'Sucursal Centro',
    'producto': 'monitores',
    'unidades': 12,
    'precio': 340000
}
print("Unidades vendidas:", venta_dicc['unidades'], "  ← acá no hay nada que recordar")

# Y el diccionario se puede corregir; la tupla no
venta_dicc['unidades'] = 15
print("\nCorregido:", venta_dicc)
````

> 💡 **Por qué esto importa para lo que viene:** cuando pasemos a pandas, un **DataFrame** es
> básicamente una lista de registros con nombre. Las columnas son las claves del diccionario y
> cada fila es una tupla de valores. Si entendés esta diferencia ahora, `df['ventas']` va a ser obvio.

## 🗂️ El `groupby` a mano: acumular por categoría

Este es **el patrón de negocio más importante** de toda la programación básica, y el puente directo
hacia pandas.

La pregunta es siempre la misma: *"tengo una lista de operaciones, ¿cuánto suma cada categoría?"*

La receta tiene tres pasos:
1. Empezar con un **diccionario vacío**.
2. Recorrer los datos.
3. Por cada uno, sumarlo en la clave que corresponda.

````python
operaciones = [
    {"ciudad": "Córdoba",  "producto": "Notebook", "monto": 890000},
    {"ciudad": "Rosario",  "producto": "Mouse",    "monto":  12500},
    {"ciudad": "Córdoba",  "producto": "Monitor",  "monto": 340000},
    {"ciudad": "Mendoza",  "producto": "Notebook", "monto": 890000},
    {"ciudad": "Rosario",  "producto": "Monitor",  "monto": 340000},
    {"ciudad": "Córdoba",  "producto": "Mouse",    "monto":  12500},
]

# Paso 1: un diccionario vacío para acumular
total_por_ciudad = {}

# Paso 2 y 3: recorrer y acumular
for op in operaciones:
    ciudad = op["ciudad"]
    # .get(clave, 0) devuelve el valor si la clave existe, y 0 si es la primera vez
    total_por_ciudad[ciudad] = total_por_ciudad.get(ciudad, 0) + op["monto"]

print("Facturación por ciudad:")
for ciudad, total in total_por_ciudad.items():
    print(f"   {ciudad:<10} $ {total:>10,.0f}")
````

````python
# Contar cuántas operaciones hubo en cada ciudad (mismo patrón, sumando 1)
cantidad_por_ciudad = {}
for op in operaciones:
    cantidad_por_ciudad[op["ciudad"]] = cantidad_por_ciudad.get(op["ciudad"], 0) + 1

print("Operaciones por ciudad:", cantidad_por_ciudad)

# Y el ticket promedio, combinando los dos diccionarios
print("\nTicket promedio:")
for ciudad in total_por_ciudad:
    promedio = total_por_ciudad[ciudad] / cantidad_por_ciudad[ciudad]
    print(f"   {ciudad:<10} $ {promedio:>10,.0f}")
````

### 🎯 Por qué esto importa

Acabás de escribir a mano lo que en pandas es **una sola línea**:

```python
df.groupby("ciudad")["monto"].sum()
```

Cuando lleguemos a esa línea, no va a ser magia: vas a saber exactamente qué está haciendo por dentro.

### ✏️ Crear tus propias funciones

Crear una función es muy simple. La forma es:

```
def nombre_de_la_funcion(arg1, arg2=valor_por_defecto, arg3...):
    ...arg1...arg2...arg3...
    return resultado
```

Tres cosas para tener en cuenta:

- **`def`** avisa que estás definiendo una función, y después va el nombre que elijas.
- Un argumento puede tener un **valor por defecto**: si no se lo pasás al llamarla, usa ese.
- **`return`** es lo que la función **devuelve**. Sin `return`, la función hace lo suyo pero no te
  entrega nada que puedas guardar en una variable.

````python
def dividir_por_dos(x):
    return x / 2

dividir_por_dos(500)
````

````python
def agregar_pausa_dramatica(palabra1, palabra2, nivel_de_drama=1):
    """ Pone la cantidad deseada de puntos suspensivos entre dos palabras.
    Muy útil si querés mucho drama. """   # con tres comillas se describe qué hace la función
    frase_dramatica = palabra1 + '...' * nivel_de_drama + palabra2
    return frase_dramatica

# como nivel_de_drama ya tiene el valor por defecto 1, podemos no pasarlo
agregar_pausa_dramatica('Voy a volver', '¡nunca!')
````

````python
# usando return, el resultado se puede guardar en una variable
ultimas_palabras = agregar_pausa_dramatica('por favor', 'traeme una factura', nivel_de_drama=3)

ultimas_palabras
````

````python
def crear_polemica_de_pizza():   # una función puede ser, simplemente, la ejecución de otras funciones
    drama_pizza = agregar_pausa_dramatica('En mi opinión', 'la piña no va en la pizza', nivel_de_drama=2)
    return drama_pizza

crear_polemica_de_pizza()
````
