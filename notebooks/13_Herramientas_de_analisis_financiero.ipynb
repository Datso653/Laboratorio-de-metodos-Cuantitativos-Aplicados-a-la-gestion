Debemos comenzar distinguiendo la tasa de interes nominal y la tasa de interes efectiva.

* ***Tasa nominal*** = `es una tasa de interés  que no toma en cuenta la frecuencia con la que se capitalizan los intereses `(es decir, si se reinvierten o no). Es decir, la nominal refiere a las normas o presentaciones.

* ***Tasa efectiva*** = `sí considera la capitalización de los intereses y muestra el costo o rendimiento real de un producto financiero en un período determinado, ya sea un préstamo o una inversión`. Es decir, cuanto obtengo de intereses absolutos (es lo que nos interesa.)

###***1) Formula de equivalencia***
¿Como pasamos de un periodo a a otro?

La fórmula general para convertir por ejemplo una TEA a TES es:

$$i(t_2) = \left(1 + i(t_1)\right)^{\frac{t_2}{t_1}} - 1$$


### Donde:
- $it2$ es la **tasa nominal buscada** (en decimales, por ejemplo: 12% = 0.12).
- $t1$ es **el periodo que tengo como dato**
- $t2$ es **el periodo que tengo que buscar**

Pasamos un dato de un periodo y lo pasamos a otro periodo


---







### ***2) Formula de Proporcion***
¿Como pasamos de EFECTIVA a NOMINAL?

La fórmula general para convertir una TN a TE es:

$$
TNA = i_{\text{efectiva, periodo}} \cdot \frac{365}{n}
$$

$$
i_{\text{efectiva, periodo}}  = TNA \cdot \frac{n}{365}
$$

### Donde:

- $TNA$ → Tasa nominal anual  
- $i_{\text{efectiva, periodo}}$ → Tasa efectiva para un período de $n$ días  
- $ n $ → Número de días del período de la tasa efectiva

### **Proporción vs. Equivalencia**

---



Para un analista de datos financieros, entender cuándo aplicar la **fórmula de proporción** y cuándo la de **equivalencia** es fundamental para no subestimar costos o sobreestimar retornos.

#### **Fórmula de Proporción**
La proporción realiza una conversión lineal del tiempo. Asume un régimen de **interés simple**, lo que significa que **no existe capitalización** (los intereses devengados no se reinvierten para generar nuevos intereses).


* **Cuándo se usa:**
  * Operaciones de muy corto plazo (usualmente menos de 30 o 90 días).
  * Instrumentos tradicionales como el descuento de cheques de pago diferido, cauciones o pagarés bursátiles.
  * Como paso previo para hallar la tasa de un período específico antes de aplicar capitalización compuesta.

#### **Fórmula de Equivalencia**
La equivalencia financiera contempla el **interés compuesto** y la reinversión de flujos (capitalización). Reconoce que el dinero tiene un valor tiempo que crece de forma exponencial, no lineal.

* **Cuándo se usa:**
  * Comparación de alternativas de inversión o financiamiento de mediano y largo plazo (ej. Crédito Prendario vs. Leasing).
  * Proyecciones de flujos de caja y cálculo del costo de capital (WACC).

#### **Diferencia Clave para la Toma de Decisiones:**
Si una PyME se financia repetidamente a tasas proporcionales de corto plazo creyendo que el costo anual es simplemente la multiplicación lineal (TNA), estará subestimando el **Costo Financiero Total Efectivo Anual (CFTEA)**, ya que la acumulación real de intereses sigue una curva exponencial regulada por la equivalencia de tasas.

## ***3) Valor futuro***

La fórmula:

$$
C_n = C_0 \cdot (1 + i)^n
$$

describe el **régimen de capitalización compuesta**. Es decir, cómo crece una inversión o deuda a lo largo del tiempo cuando los intereses se reinvierten al final de cada período.

### Desglose:

- $ C_0 $: capital inicial, es decir, lo que invertís o prestás hoy.
- $ i $: tasa de interés por período (por ejemplo, mensual si capitalizás mensualmente).
- $ n $: número total de períodos (por ejemplo, meses).
- $C_n $: capital final acumulado después de \( n \) períodos, con interés compuesto.

---






## ***4) Valor Presente***

La fórmula:

$$
C_0 = \frac{C_n}{(1 + i)^n}
$$

Es básicamente la inversa del cálculo de valor futuro: en lugar de proyectar cuánto tendrás más adelante, traés un monto futuro a valor de hoy. Esto se llama descontar o actualizar un valor.



### Desglose:
- C_0: Capital inicial o valor presente, lo que invertís o prestás hoy.
- C_n: Capital final o valor futuro, monto acumulado al final.
- i: Tasa de interés por período (por ejemplo, mensual si capitalizás mensualmente).
- n: Número total de períodos (meses, años, etc.).

---

### **Valor Futuro y Valor Presente (Descuento) en la Gestión PyME**

En la administración financiera diaria de una PyME, el dinero no tiene un valor estático; su valor cambia a lo largo del tiempo debido a la tasa de interés y al costo de oportunidad del capital. Entender este fenómeno requiere dominar dos procesos inversos y fundamentales:

#### **La Capitalización (Llevar al Futuro)**
Es el proceso de proyectar un capital actual hacia el futuro, sumando los intereses que este genera período tras período. En el régimen de **capitalización compuesta**, los intereses generados se reinvierten (se adicionan al capital original), de modo que en el siguiente período generan nuevos intereses de forma exponencial.



#### **La Actualización o Descuento (Traer al Presente)**
La **actualización** o descuento es el proceso inverso. Permite determinar cuánto vale hoy un flujo de dinero que se recibirá o pagará en el futuro. Es una herramienta clave para calcular el costo de oportunidad y evaluar si una promesa de pago futura (como un cheque diferido o una factura de crédito) justifica el desembolso o la espera de hoy.



### **Variables para una PyME:**
* **$C_0$ (Capital Inicial o Valor Presente):** Representa la liquidez inmediata. Es el dinero en efectivo que la empresa tiene disponible hoy (por ejemplo, para pagar salarios o comprar insumos) o el monto neto que recibe al descontar un cheque de pago diferido en el mercado financiero.
* **$C_n$ (Capital Final o Valor Futuro):** Es el valor nominal de un flujo al final del período $n$. Por ejemplo, el monto escrito en un cheque diferido que se cobrará dentro de 60 días, o el valor final de un fondo de reserva colocado en un Plazo Fijo.
* **$i$ (Tasa de Interés o de Descuento por Período):** Es la tasa que mide el rendimiento de una inversión (si capitalizamos) o el costo de adelantar el dinero (si descontamos). Para una PyME, esta tasa suele representar su costo de capital (WACC) o la tasa exigida para aceptar un proyecto.
* **$n$ (Número de Períodos):** El tiempo durante el cual el dinero cambia de valor (expresado en días, meses o años, debiendo coincidir siempre con la periodicidad de la tasa $i$).

### **Ejemplos de uso**
* **Evaluación de Cobros Diferidos:** Si un cliente propone pagar una venta a 90 días con un recargo, la PyME debe calcular el *Valor Presente* de ese cobro futuro usando su tasa de costo financiero. Si el valor presente es menor que el precio de venta de contado, la empresa estará perdiendo dinero en términos reales.
* **Costo de Oportunidad:** Permite evaluar si es conveniente conservar un activo financiero hasta su vencimiento o liquidarlo anticipadamente para inyectar liquidez en el ciclo operativo del negocio (compra de mercadería con alta rotación).

```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
```

```python
# 1) Fórmula de Proporción
def proporcion(TNA, n):
    return TNA * (n / 365)

# 2) Fórmula de Equivalencia
def equivalencia(i_t1, t2, t1):
    return (1 + i_t1) ** (t2 / t1) - 1

# 3) Capitalización Compuesta
def capitalizacion_compuesta(C0, i, n):
    return C0 * (1 + i) ** n

# 4) Valor presente
def valor_presente(Cn, i, n):
    return Cn/((1 + i) ** n )
```

```python
def tna_a_te(tna: float, m: int, periodo: str = "mensual") -> float:
    """
    Convierte una Tasa Nominal Anual (TNA) a una Tasa Efectiva.

    Parámetros
    ----------
    tna : float
        Tasa Nominal Anual en decimal (ej: 0.40 para 40%).
    m : int
        Frecuencia de capitalización por año:
        - 1   → anual
        - 2   → semestral
        - 4   → trimestral
        - 12  → mensual
        - 365 → diaria
    periodo : str
        Período de la tasa efectiva que querés obtener:
        "anual", "semestral", "trimestral", "mensual", "diaria".

    Retorna
    -------
    float
        Tasa efectiva del período pedido, en decimal.
    """
    periodos_por_año = {
        "anual": 1,
        "semestral": 2,
        "trimestral": 4,
        "mensual": 12,
        "diaria": 365,
    }

    if periodo not in periodos_por_año:
        raise ValueError(f"Período inválido: {periodo}")

    n = periodos_por_año[periodo]

    # Tasa efectiva del período pedido
    te = (1 + tna / m) ** (m / n) - 1
    return te
```

No considera que los intereses generados en un período más corto (por ejemplo, cada día) se suman al capital y comienzan a generar sus propios intereses. Por lo tanto, la tasa que resulta de esta fórmula no es la tasa real (efectiva) del período.

#### **Caso 1: El verdadero costo de pagar el mínimo de la Tarjeta de Crédito**

Imaginá que un banco te cobra una **TNA del 60%** por financiar el saldo de tu tarjeta de crédito con capitalización mensual. Sin embargo, para obtener el **Costo Financiero Total (CFT)**, debemos sumarle un **21% de IVA sobre los intereses** y un seguro de vida obligatorio mensual del **0.1% sobre el saldo financiado**.

Vamos a calcular:
1. La Tasa Efectiva Mensual (TEM) básica.
2. La TEM cargada con IVA.
3. La Tasa Efectiva Anual (TEA) y el Costo Financiero Total Efectivo Anual (CFTEA) aproximado.

```python
# ---------- Datos ----------
TNA = 0.60              # 60% nominal anual
iva = 0.21              # 21% sobre intereses
seguro_mensual = 0.001  # 0.1% mensual sobre saldo

# ---------- 1) TEM básica ----------
# Capitalización MENSUAL → la TEM sale directo de TNA/12
tem_basica = TNA / 12

# ---------- 2) TEM con IVA ----------
# El IVA se aplica SOLO sobre los intereses
tem_con_iva = tem_basica * (1 + iva)

# ---------- 3) TEM total (interés + IVA + seguro) ----------
tem_total_mensual = tem_con_iva + seguro_mensual

# ---------- 4) TEA y CFTEA ----------
# Capitalización mensual → 12 períodos por año
tea_basica = (1 + tem_basica) ** 12 - 1
cftea      = (1 + tem_total_mensual) ** 12 - 1


print(f"TEM Básica (sin impuestos)........: {tem_basica*100:.3f}%")
print(f"TEM con IVA.......................: {tem_con_iva*100:.3f}%")
print(f"TEM Total con IVA y Seguro........: {tem_total_mensual*100:.3f}%")
print(f"TEA (solo interés)................: {tea_basica*100:.2f}%")
print(f"CFTEA (costo real de financiación): {cftea*100:.2f}%")
```

#### **Caso 3: Plazo Fijo vs. Inflación (La Tasa de Interés Real)**

Un ahorrista coloca **$1.000.000** en un plazo fijo a 30 días que rinde una **TNA del 45%**. Durante ese mismo mes, la inflación registrada por el IPC fue del **3.5%**.

Vamos a calcular:
1. Cuánto dinero retira al final de los 30 días.
2. La tasa de rendimiento mensual obtenida.
3. La **Tasa de Interés Real** según la fórmula de Fisher:

$$1 + i_{real} = \frac{1 + i_{nominal}}{1 + \text{inflación}}$$


---



```python
# ---------- Datos ----------
capital_inicial = 1_000_000
TNA_pf = 0.45               # TNA publicada (base 365, convención BCRA)
plazo_dias = 30             # plazo fijo a 30 días
inflacion_mes = 0.035       # 3.5% mensual (IPC)

# ---------- 1) Tasa efectiva del período (30 días) ----------
# Convención BCRA para plazos fijos: TNA × (plazo/365)
i_periodo = TNA_pf * (plazo_dias / 365)

# ---------- 2) Capital al vencimiento (en pesos corrientes) ----------
capital_final = capital_inicial * (1 + i_periodo)
intereses = capital_final - capital_inicial

# ---------- 3) Tasa Real del período (Fisher) ----------
i_real = ((1 + i_periodo) / (1 + inflacion_mes)) - 1

# ---------- 4) Capital en pesos CONSTANTES (poder de compra) ----------
# Cuánto vale el capital final medido en pesos del momento inicial
capital_final_real = capital_final / (1 + inflacion_mes)
ganancia_real_ = capital_final_real - capital_inicial

# ---------- 5) TEA del plazo fijo (para comparar con inflación anual) ----------
# Renovando el PF cada 30 días durante 365 días:
tea_pf = (1 + i_periodo) ** (365 / plazo_dias) - 1
tea_inflacion = (1 + inflacion_mes) ** 12 - 1


print("  PLAZO FIJO vs. INFLACIÓN")

print(f"  Capital inicial...............: ${capital_inicial:,.2f}")
print(f"  TNA plazo fijo................: {TNA_pf*100:.2f}%")
print(f"  Plazo.........................: {plazo_dias} días")
print(f"  Inflación del mes.............: {inflacion_mes*100:.2f}%")

print(f"  Tasa efectiva del período.....: {i_periodo*100:.4f}%")
print(f"  Intereses ganados.............: ${intereses:,.2f}")
print(f"  Capital final (pesos corr.)...: ${capital_final:,.2f}")

print(f"  Tasa REAL del período.........: {i_real*100:.4f}%")
print(f"  Capital final (pesos const.)..: ${capital_final_real:,.2f}")
print(f"  Ganancia/pérdida REAL.........: ${ganancia_real_:,.2f}")

print(f"  TEA plazo fijo (renovando)...: {tea_pf*100:.2f}%")
print(f"  TEA inflación (anualizada)...: {tea_inflacion*100:.2f}%")


if i_real > 0:
    print("Ganaste poder adquisitivo.")
else:
    print("Perdiste poder adquisitivo frente a la inflación.")
```

#### **Caso 4: Costo de Capital de una Pyme (WACC Simplificado)**

Una pequeña empresa necesita financiar un proyecto de inversión de 10.000.000 de pesos.
* Financiará 6.000.000 mediante un préstamo bancario a una **Tasa Efectiva Anual (TEA) del 50%**.
* El resto (**$4.000.000**) lo aportarán los socios, quienes exigen un retorno mínimo (**Costo del Capital Propio**) del **65% anual**.
* La alícuota del impuesto a las ganancias de la empresa es del **30%** (los intereses de la deuda bancaria son deducibles de impuestos, lo que genera un **escudo fiscal**).

Calculamos el Costo Promedio Ponderado de Capital (WACC)


---



```python
# ---------- Datos ----------
monto_total    = 10_000_000
deuda          = 6_000_000
capital_propio = 4_000_000

ke       = 0.65    # Costo de capital propio (socios)
kd       = 0.50    # TEA de la deuda (ya es efectiva)
tax_rate = 0.30    # Impuesto a las Ganancias

# ---------- 1) Ponderaciones ----------
wd = deuda / monto_total
we = capital_propio / monto_total

# ---------- 2) Costo de la deuda neto de impuestos ----------
# Escudo fiscal: los intereses son deducibles → el fisco "paga" parte
kd_neto = kd * (1 - tax_rate)

# ---------- 3) WACC ----------
wacc = (we * ke) + (wd * kd_neto)

# ---------- 4) Ahorro fiscal anual en pesos ----------
intereses_anuales = deuda * kd
ahorro_fiscal = intereses_anuales * tax_rate
costo_deuda_neto_ = intereses_anuales - ahorro_fiscal


print("=" * 62)
print("  WACC — COSTO DE CAPITAL DE LA PYME")
print("=" * 62)
print(f"  Inversión total...............: ${monto_total:,.0f}")
print(f"  Deuda bancaria (TEA {kd*100:.0f}%).......: ${deuda:,.0f}")
print(f"  Capital propio (ke {ke*100:.0f}%)........: ${capital_propio:,.0f}")
print("-" * 62)
print(f"  Ponderación Deuda (Wd)........: {wd*100:.1f}%")
print(f"  Ponderación Capital (We)......: {we*100:.1f}%")
print("-" * 62)
print(f"  Costo deuda antes de imp.....: {kd*100:.2f}%")
print(f"  Escudo fiscal (t = {tax_rate*100:.0f}%)......: -{kd*tax_rate*100:.2f}%")
print(f"  Costo deuda después de imp...: {kd_neto*100:.2f}%")
print("-" * 62)
print(f"  Aporte capital propio al WACC: {we*ke*100:.2f}%")
print(f"  Aporte deuda al WACC.........: {wd*kd_neto*100:.2f}%")
print("-" * 62)
print(f"  WACC.........................: {wacc*100:.2f}%")
print("-" * 62)
print(f"  Intereses anuales.............: ${intereses_anuales:,.0f}")
print(f"  Ahorro fiscal por escudo......: ${ahorro_fiscal:,.0f}")
print(f"  Costo neto de la deuda........: ${costo_deuda_neto_:,.0f}")
print("=" * 62)
print(f"\n  ➜ El proyecto debe rendir ≥ {wacc*100:.2f}% anual para crear valor.")
```

## 5) Valor Actual Neto (VAN)

El Valor Actual Neto (VAN) es uno de los indicadores más importantes para evaluar la viabilidad económica de un proyecto de inversión. `Se calcula trayendo a valor presente todos los flujos de caja futuros esperados del proyecto y restándole la inversión inicial.`

La fórmula general del VAN es:

$$
VAN = \sum_{t=0}^{n} \frac{F_t}{(1+r)^t} - I_0
$$

Donde:

- $ VAN $: Valor Actual Neto
- $ F_t $: Flujo de caja en el período \( t \)
- $ r $: Tasa de descuento o tasa de rendimiento requerida (costo de capital)
- $ t $: Período de tiempo
- $ n $: Número total de períodos
- $ I_0 $: Inversión inicial (que generalmente se representa como un flujo de caja negativo en el período 0)

**Interpretación del VAN:**

- Si **VAN > 0**: El proyecto es rentable y se espera que genere valor por encima de la tasa de rendimiento requerida. Conviene aceptar el proyecto.
- Si **VAN < 0**: El proyecto no es rentable y se espera que destruya valor. No conviene aceptar el proyecto.
- Si **VAN = 0**: El proyecto es indiferente. Genera exactamente la tasa de rendimiento requerida, pero no crea valor adicional.


# ¿Cual es la diferencia con el valor presente?

* El valor presente (VP) es el valor actual de uno o varios flujos futuros descontados a una tasa determinada, mientras que el valor actual neto (VAN) es ese valor presente total menos la inversión inicial. En otras palabras, el VP te dice cuánto valen hoy los flujos futuros, y el VAN te indica si el proyecto crea o destruye valor

---

Ahora, veamos un ejemplo práctico en código.


---



```python
#lista
flujos_caja = {
    'Año': [0, 1, 2, 3, 4, 5],
    'Flujo': [-10000, 3000, 3500, 4000, 4000, 2000]
}

# Creo df
df_flujos = pd.DataFrame(flujos_caja)

df_flujos
```

```python
# Definir la tasa de descuento para ver cuanto saca periodo a periodo
tasa_descuento = 0.50

# valor presente
df_flujos['Valor Presente'] = df_flujos['Flujo'] / (1 + tasa_descuento) ** df_flujos['Año'] # por cada flujo lo descuenta por una tasa de interes

# Calcular el VAN sumando los valores presentes de todos los flujos
van_calculado = df_flujos['Valor Presente'].sum()

# Mostrar el DataFrame con los valores presentes
print("Tabla de Flujos de Caja y sus Valores Presentes:")
display(df_flujos)

# Mostrar el resultado del VAN
print(f"\nEl Valor Actual Neto (VAN) del proyecto es: ${van_calculado:.2f}")
```

```python
#Guardemos una funcion donde calcule VPN
def calcular_vpn(flujos, tasa):
    ##Formula que calcula VPN, toma un array y una tasa la cual debemos definir con ese mismo nombre)##

    df = pd.DataFrame(flujos)
    df['VP'] = df['Flujo'] / (1 + tasa) ** df['Año']
    vpn = df['VP'].sum()
    return df, vpn
```

### ***6) Tasa Interna de Retorno (TIR)***

La Tasa Interna de Retorno (TIR) es la tasa de descuento que hace que el Valor Actual Neto (VAN) de todos los flujos de caja de un proyecto de inversión sea igual a cero. Representa la rentabilidad intrínseca del proyecto.

**La fórmula es implícita y generalmente requiere métodos iterativos para su cálculo:**

$$
\sum_{t=0}^{n} \frac{F_t}{(1+TIR)^t} = 0
$$

Donde:

-$ F_t $: Flujo de caja en el período \( t \)
- $ TIR $: Tasa Interna de Retorno
- $ t $: Período de tiempo
- $ n $: Número total de períodos

**Interpretación de la TIR:**

- Si **TIR > r** (Tasa de descuento o costo de capital): El proyecto es rentable y conviene aceptarlo.
- Si **TIR < r**: El proyecto no es rentable y no conviene aceptarlo.
- Si **TIR = r**: El proyecto es indiferente.

```python
def calcular_tir(flujos):
    """Calcula la TIR por bisección sobre los flujos."""
    flujos_arr = np.array(flujos['Flujo'], dtype=float)
    años = np.array(flujos['Año'], dtype=float)

    def vpn_en(tasa):
        return np.sum(flujos_arr / (1 + tasa) ** años)

    lo, hi = -0.99, 10.0
    for _ in range(200):
        mid = (lo + hi) / 2
        if vpn_en(mid) > 0:
            lo = mid
        else:
            hi = mid
    return (lo + hi) / 2

```

### ***7) Tasa Interna de Retorno Modificada (TIRM)***

La Tasa Interna de Retorno Modificada (TIRM) aborda una de las críticas de la TIR tradicional: la suposición de que los flujos de caja intermedios se reinvierten a la propia TIR. La TIRM asume que los flujos de caja positivos se reinvierten a una tasa de reinversión específica (generalmente el costo de capital), y los flujos de caja negativos se descuentan al costo de financiamiento.

**Fórmula de la TIRM:**

$$
TIRM = \left( \frac{VF_{positivos}}{VP_{negativos}} \right)^{\frac{1}{n}} - 1
$$

Donde:

- $ VF_{positivos} $: Valor futuro de los flujos de caja positivos (reinvertidos a la tasa de reinversión).
- $ VP_{negativos} $: Valor presente de los flujos de caja negativos (descontados al costo de financiamiento).
- $ n $: Número total de períodos.

**Interpretación de la TIRM:**

- Si **TIRM > r** (Tasa de reinversión/costo de capital): El proyecto es rentable y conviene aceptarlo.
- Si **TIRM < r**: El proyecto no es rentable y no conviene aceptarlo.
- Si **TIRM = r**: El proyecto es indiferente.

```python
def calcular_tirm(flujos, tasa_reinversion, tasa_financiamiento=None):
    df = pd.DataFrame(flujos).copy()
    if tasa_financiamiento is None:
        tasa_financiamiento = tasa_reinversion

    n = df['Año'].max()
    positivos = df[df['Flujo'] > 0]
    negativos = df[df['Flujo'] < 0]

    # VF de los flujos positivos (capitalizados hasta año n)
    VF_positivos = sum(
        f * (1 + tasa_reinversion) ** (n - a)
        for a, f in zip(positivos['Año'], positivos['Flujo'])
    )

    # VP de los flujos negativos (en valor absoluto)
    VP_negativos_abs = abs(sum(
        f / (1 + tasa_financiamiento) ** a
        for a, f in zip(negativos['Año'], negativos['Flujo'])
    ))

    tirm = (VF_positivos / VP_negativos_abs) ** (1 / n) - 1
    return tirm, VF_positivos, VP_negativos_abs


```
