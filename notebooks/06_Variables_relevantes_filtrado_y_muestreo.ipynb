Muestreo Estratificado por Fraude

El muestreo estratificado es útil cuando la variable objetivo (en este caso, 'Fraud') está desequilibrada. Nos permite asegurar que cada 'estrato' o grupo de la variable objetivo esté representado proporcionalmente o con un número específico en la submuestra. Aquí, aseguraremos que tanto las transacciones fraudulentas como las no fraudulentas estén presentes en la submuestra.

```python
# Primero, vamos a verificar la distribución de la columna 'Fraud' en el dataset original.
display(data['Fraud'].value_counts())
```

```python
# Realizamos un muestreo estratificado para tener una representación equilibrada de 'Fraud'.
# Por ejemplo, tomaremos 1000 muestras de cada categoría (0 y 1) si hay suficientes, o todas las disponibles.

fraud_sample = data[data['Fraud'] == 1].sample(n=min(1000, data['Fraud'].value_counts()[1]), random_state=42)
non_fraud_sample = data[data['Fraud'] == 0].sample(n=min(1000, data['Fraud'].value_counts()[0]), random_state=42)

submuestra_estratificada = pd.concat([fraud_sample, non_fraud_sample])

print("Dimensiones de la submuestra estratificada:", submuestra_estratificada.shape)
display(submuestra_estratificada['Fraud'].value_counts())
display(submuestra_estratificada.head())
```

```python
submuestra_estratificada=submuestra_estratificada.reset_index()
submuestra_estratificada
```

Muestreo por Montos y Fraude (High-Value Transactions Sampling)

Este método se enfoca en transacciones con montos inusualmente altos, ya que el fraude a menudo implica grandes sumas de dinero. Realizaremos un muestreo de estas transacciones de alto valor, prestando especial atención a cómo se distribuye la etiqueta de fraude dentro de este subconjunto. Esto nos permitirá examinar de cerca los patrones de fraude en los casos más costosos.

```python
# Identificamos un umbral para las transacciones de alto valor (por ejemplo, el 95 percentil del monto)
umbral_alto_valor = submuestra_estratificada['Amount'].quantile(0.95)
print(f"Umbral de alto valor (percentil 95): {umbral_alto_valor:.2f}")
```

```python
# Filtramos las transacciones de alto valor
transacciones_alto_valor = submuestra_estratificada[submuestra_estratificada['Amount'] >= umbral_alto_valor].reset_index()
transacciones_alto_valor
```

### Muestreo por Conglomerados (Cluster Sampling)

El muestreo por conglomerados implica dividir la población en grupos (conglomerados) y luego seleccionar aleatoriamente algunos de estos conglomerados, muestreando a todos los miembros de los conglomerados seleccionados. En este caso, podemos considerar 'Cardnum' (número de tarjeta) como un conglomerado. Seleccionaremos aleatoriamente algunos números de tarjeta y obtendremos todas sus transacciones.

```python
# Identificamos los conglomerados (Cardnum únicos) de las transacciones de alto valor
card_numbers = transacciones_alto_valor['Cardnum'].unique()
display(card_numbers)
```

```python
# Identificamos los Cardnum que tienen más de una transacción en el dataset de transacciones de alto valor.
# Esto busca Cardnums con múltiples transacciones fraudulentas de alto valor
cardnum_counts = transacciones_alto_valor['Cardnum'].value_counts()
suspicious_cardnums_multi_trans = cardnum_counts[cardnum_counts > 5].index.unique()

# Filtramos el DataFrame de transacciones de alto valor para obtener todas las transacciones de estos Cardnum "sospechosos".
submuestra_cluster_sospechosa = transacciones_alto_valor[transacciones_alto_valor['Cardnum'].isin(suspicious_cardnums_multi_trans)]

print(f"Dimensiones de la submuestra por conglomerados (Cardnum con múltiples transacciones fraudulentas de alto valor): {submuestra_cluster_sospechosa.shape}")
print(f"Número de Cardnum únicos seleccionados (con múltiples transacciones fraudulentas de alto valor): {submuestra_cluster_sospechosa['Cardnum'].nunique()}")
display(submuestra_cluster_sospechosa)
```

```python
#importamos el conjunto de datos que vamos a procesar
url = "https://raw.githubusercontent.com/sanchezgaviermatias/Curso_Python-/master/2%20-%20Pandas/Pokemon.csv"
#asignamos un nombre y creamos un DataFrame con los datos sobre los cuales trabajaremos
archivo_datos = pd.read_csv(url)
archivo_datos
```

```python
#Vemos número de filas, número y tipo de columnas,índice y memoria usada
archivo_datos.info()
```

```python
#Buscamos los datos estadísticos del archivo
#elementos por columna, media, desvío, percentiles 25,50,75,máximos y mínimos
archivo_datos.describe().transpose()
```

```python
# Cambiamos los nombres de Columnas
#usamos diccionarios
archivo_datos.rename(columns = {"Name":"Nombre","Speed":"Velocidad","Generation":"Generación",'Type 1':'Poder 1', "Type 2": "Poder 2", "HP": "Salud"}, inplace=True)
archivo_datos
```

```python
#buscamos varias columnas
archivo_datos.loc[:,["Attack","Defense","Sp. Atk","Sp. Def"]]
```

```python
#vemos los datos únicos de una columna
archivo_datos["Poder 1"].unique()
```

```python
archivo_datos["Velocidad"].unique()
```

```python
#creamos una nueva variable a partir de otras
#generamos una columna que clasifique el tiempo de reacción en lenta,mediana,veloz,muy veloz

archivo_datos.loc[archivo_datos["Velocidad"] <=50 , 'Reacción'] = "Lenta"

archivo_datos.loc[(51<=archivo_datos["Velocidad"]) & (archivo_datos["Velocidad"] <=100) , 'Reacción'] = "Mediana"

archivo_datos.loc[(101<=archivo_datos["Velocidad"]) & (archivo_datos["Velocidad"] <=130) , 'Reacción'] = "Veloz"

archivo_datos.loc[archivo_datos["Velocidad"]>=131 , 'Reacción'] = "Muy Veloz"
archivo_datos
```

```python
## Operaciones entre Columnas

archivo_datos["Poder de Ataque"] = archivo_datos["Salud"]*(1/3) + archivo_datos["Attack"]*(1/3)+archivo_datos["Sp. Atk"]*(1/3)
archivo_datos["Poder de Ataque"] = round(archivo_datos["Poder de Ataque"])

archivo_datos["Poder de Defenza"] = archivo_datos["Salud"]*(1/3) + archivo_datos["Defense"]*(1/3)+archivo_datos["Sp. Def"]*(1/3)
archivo_datos["Poder de Defenza"] = round(archivo_datos["Poder de Defenza"])
archivo_datos
```

```python
# Podemos aplicar funciones a las columnas

def pokemon_poder(poder):
    """Imprime los pokemon con un determinado poder seleccionado"""
    poder =  (archivo_datos["Poder 1"].isin([poder]))
    datos_selec=archivo_datos.loc[ poder, : ]

    return datos_selec

pokemon_poder("Grass")
```

```python
#Datos faltantes
#Métodos de resolución de NaN
#Paso 1- vemos donde hay datos faltantes
archivo_datos.isnull().sum()
#muestra que campo tiene datos faltantes y los cuantifica
```

```python
#Paso 2- resolver los datos faltantes
archivo_datos["Poder 2"]=archivo_datos["Poder 2"].fillna("Ninguno")
#Paso 3- chequeamos resultados
archivo_datos.isnull().sum()
```
