Planteo del problema


---

Una cadena de comida rápida planea agregar un nuevo artículo a su menú. Sin embargo, todavía están indecisos entre tres posibles campañas de marketing para promocionar el nuevo producto. Para determinar qué promoción tiene el mayor efecto sobre las ventas, el nuevo artículo se introduce en ubicaciones de varios mercados seleccionados al azar. Se utiliza una promoción diferente en cada ubicación y las ventas semanales del nuevo artículo se registran durante las primeras cuatro semanas.

Dataset:

MarketID: identificador único para el mercado

MarketSize: tamaño del área de mercado por ventas

LocationID: identificador único para la ubicación de la tienda

AgeOfStore: antigüedad de la tienda en años

Promoción: una de las tres promociones que se probaron

Semana: una de las cuatro semanas en las que se realizaron las promociones.

SalesInThousands: monto de ventas para un ID

Objetivo:
el objetivo en este caso es determinar cuál de las tres campañas de marketing genera las mayores ventas para el nuevo elemento del menú.

Análisis de datos:

1- Analizamos los datos comparando los montos de venta de cada promoción.

2- Calculamos las ventas promedio por semana para cada promoción en todas las ubicaciones.

3- Analizamos los datos segmentándolos según tamaño del área del mercado o antigüedad para comprender cómo los diferentes factores pueden influir en el éxito de cada promoción.

4- Filtramos los datos por Promoción (Promoción 1, Promoción 2 y Promoción 3).
Para cada promoción, sume la columna SalesInThousands.
Divida las ventas totales por la cantidad de semanas para obtener el promedio de ventas semanales para cada promoción.

5-Evaluamos los resultados: compare las ventas semanales promedio de cada promoción para determinar cuál tuvo el mejor desempeño. Considere cualquier información adicional que haya obtenido al segmentar los datos según MarketSize o AgeOfStore.

```python
# Visualice las ventas promedio por AgeOfStore y Promoción (usando grupos de edad)
data["AgeBin"] = pd.cut(data["AgeOfStore"], bins=[0, 5, 10, 15, 20, 25, 30], labels=["0-5", "6-10", "11-15", "16-20", "21-25", "26-30"])
edades = data.groupby(["AgeBin", "Promotion"])["SalesInThousands"].mean().reset_index()
sb.barplot(x="AgeBin", y="SalesInThousands", hue="Promotion", data=edades)
plt.title("Ventas promedio por antigüedad de la tienda y promoción")
plt.xlabel("Grupos de edades")
plt.ylabel("Ventas en miles usd")
plt.show()
```

```python
#vemos las ventas semanales promedio por promoción y evaluamos
ventas = data.groupby(["Promotion", "week"])["SalesInThousands"].sum().reset_index()
ventas_semanales_promedio = ventas.groupby("Promotion")["SalesInThousands"].mean().reset_index()
sb.barplot(x="Promotion", y="SalesInThousands", data=ventas_semanales_promedio)
plt.title("Ventas semanales promedio por promoción")
plt.xlabel("Promoción")
plt.ylabel("Ventas en miles usd")
plt.show()
```

Caso 3 - Vamos a generar una base Simulada de un supermercado

---







```python
# Crear datos ficticios para un archivo Supermarket_Sales.csv
np.random.seed(42) # para reproducibilidad

productos = ['Leche', 'Yogur', 'Queso', 'Carne Res', 'Pollo', 'Pescado', 'Manzanas', 'Bananas', 'Lechuga', 'Pan Blanco', 'Galletas']
categorias_map = {
    'Leche': 'Lácteos',
    'Yogur': 'Lácteos',
    'Queso': 'Lácteos',
    'Carne Res': 'Carnes',
    'Pollo': 'Carnes',
    'Pescado': 'Carnes',
    'Manzanas': 'Frutas y Verduras',
    'Bananas': 'Frutas y Verduras',
    'Lechuga': 'Frutas y Verduras',
    'Pan Blanco': 'Panadería',
    'Galletas': 'Panadería'
}

regiones = ['Norte', 'Sur', 'Este', 'Oeste']

# Generar nombres de productos aleatorios
nombres_productos_generados = np.random.choice(productos, 1000)

# Asignar categorías basadas en el mapeo
categorias_generadas = [categorias_map[prod] for prod in nombres_productos_generados]

data = {
    'nombre_producto': nombres_productos_generados,
    'categoria': categorias_generadas,
    'fecha_venta': pd.to_datetime('2025-01-01') + pd.to_timedelta(np.random.randint(0, 365, 1000), unit='D'),
    'unidades_vendidas_diarias': np.random.randint(10, 200, 1000),
    'precio_unitario': np.round(np.random.uniform(1.5, 50, 1000), 2),
    'costo_unitario': np.round(np.random.uniform(0.5, 30, 1000), 2),
    'region': np.random.choice(regiones, 1000)
}
df_supermarket = pd.DataFrame(data)
df_supermarket
```

Ahora generamos más variables a partir de otras

```python
df_supermarket['ingreso_diario'] = df_supermarket['unidades_vendidas_diarias'] * df_supermarket['precio_unitario']
df_supermarket['margen_ganancia'] = (df_supermarket['precio_unitario'] - df_supermarket['costo_unitario']) * df_supermarket['unidades_vendidas_diarias']
df_supermarket['margen_ganancia_porcentual'] = (df_supermarket['margen_ganancia'] / df_supermarket['ingreso_diario']).fillna(0)


df_supermarket['cambio_ventas_semanal'] = np.round(np.random.uniform(-15, 20, 1000), 2) # % de cambio
df_supermarket
```

El gráfico anterior no es visualmente claro, vamos a mejorarlo tomando en cuenta la naturaleza de la variable fecha...

```python
# Extraer el mes y el año de la fecha de venta
df_supermarket['año_mes'] = df_supermarket['fecha_venta'].dt.to_period('M')

# Agrupar por mes y sumar el ingreso diario
ingreso_mensual_total = df_supermarket.groupby('año_mes')['ingreso_diario'].sum().reset_index()

# Convertir 'año_mes' a string para un mejor etiquetado en el eje x
ingreso_mensual_total['año_mes'] = ingreso_mensual_total['año_mes'].astype(str)

plt.figure(figsize=(18, 8))
sb.lineplot(x='año_mes', y='ingreso_diario', data=ingreso_mensual_total, color='blue')
plt.title('Tendencia Mensual del Ingreso Diario Total')
plt.xlabel('Mes y Año')
plt.ylabel('Ingreso Diario Mensual Total')
plt.grid(True, linestyle='--', alpha=0.7)
plt.xticks(rotation=45)
plt.ticklabel_format(style='plain', axis='y')
plt.tight_layout()
plt.show()
```

```python
# 7 - Gráfico de barras, productos con mayor y menor ganancia

# 8. Gráfico de Barras Horizontales: Productos con Mayor y Menor Beneficio
productos_beneficio = df_supermarket.groupby('nombre_producto')['margen_ganancia'].sum().reset_index()

# Top 5 productos con mayor beneficio
top_5_beneficio = productos_beneficio.nlargest(5, 'margen_ganancia').copy()
top_5_beneficio['Tipo_Beneficio'] = 'Mayor Beneficio'

# Top 5 productos con menor beneficio
bottom_5_beneficio = productos_beneficio.nsmallest(5, 'margen_ganancia').copy()
bottom_5_beneficio['Tipo_Beneficio'] = 'Menor Beneficio'

# Combinar ambos para la visualización
productos_interes = pd.concat([top_5_beneficio, bottom_5_beneficio])

plt.figure(figsize=(12, 8))
sb.barplot(x='margen_ganancia', y='nombre_producto', hue='Tipo_Beneficio', data=productos_interes, palette={'Mayor Beneficio': 'green', 'Menor Beneficio': 'red'})
plt.title('Productos con Mayor y Menor Margen de Ganancia')
plt.xlabel('Margen de Ganancia')
plt.ylabel('Nombre del Producto')
plt.ticklabel_format(style='plain', axis='x')
plt.legend(title='Tipo de Beneficio')
plt.show()
```
